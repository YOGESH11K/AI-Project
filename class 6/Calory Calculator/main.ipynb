```python
print("🔥 Calorie Calculator")
print("---------------------")

name = input("Enter your name: ")

age = int(input("Enter your age: "))
weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in cm: "))

# Simple calorie estimation
calories = 10 * weight + 6.25 * height - 5 * age

print("\nHello", name)
print("Estimated calories:", round(calories, 2), "calories")

# Simple activity suggestion
activity = input("Are you active today? (yes/no): ").lower()

if activity == "yes":
    print("🏃 You are active today!")
else:
    print("🚶 Try to stay active with some normal daily movement.")

```
