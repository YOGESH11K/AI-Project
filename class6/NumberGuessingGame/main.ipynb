```python
import random

while True:
    number = random.randint(1, 10)
    attempts = 0
    max_attempts = 3

    print("\n🎯 Number Guessing Game")
    print("Maine 1 se 10 ke beech ek number choose kiya hai.")
    print("Tumhare paas 3 chances hain!")

    while attempts < max_attempts:
        guess = int(input("Apna guess enter karo: "))
        attempts += 1

        if guess < number:
            print("📈 Too low!")
        elif guess > number:
            print("📉 Too high!")
        else:
            print(f"🎉 Correct! Number {number} tha.")
            print(f"✅ Tumne {attempts} attempts mein guess kar liya!")
            break
    else:
        print(f"\n❌ 3 attempts khatam!")
        print(f"Number {number} tha.")

    again = input("\n🔄 Dobara try karna hai? (yes/no): ").lower()

    if again != "yes":
        print("👋 Thanks for playing!")
        break
```
